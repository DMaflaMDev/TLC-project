### Boxplots

Perform a check for outliers on relevant columns such as trip distance and trip duration. Remember, some of the best ways to identify the presence of outliers in data are box plots and histograms. 

**Note:** Remember to convert your date columns to datetime in order to derive total trip duration.  

```python
# Convert data columns to datetime
df['tpep_pickup_datetime'] = pd.to_datetime(df['tpep_pickup_datetime'])
df['tpep_dropoff_datetime'] = pd.to_datetime(df['tpep_dropoff_datetime'])
```

**trip distance**

```python
# Create box plot of trip_distance
plt.figure(figsize=(10,3))
sns.boxplot(data=df, x='trip_distance', fliersize=1)
plt.title('Trip distance distribution', fontsize=18)
plt.xlabel('Trip distance in miles')
```

```python
# Create histogram of trip_distance
plt.figure(figsize=(12,6))
sns.histplot(data=df, x='trip_distance', binwidth=1)
plt.title('Trip distance histogram', fontsize=18)
plt.xlabel('Trip distance in miles')
plt.ylabel('Times a trip distance value is in the data')
```

**total amount**

```python
# Create box plot of total_amount
plt.figure(figsize=(15,3))
sns.boxplot(data=df, x='total_amount', fliersize=1)
plt.title('Total amount distribution', fontsize=18)
plt.xlabel('Total amount in dollars')
```

```python
# Create histogram of total_amount
plt.figure(figsize=(12,6))
sns.histplot(data=df, x='total_amount', bins=range(-125,1250,10))
plt.title('Total amount histogram', fontsize=18)
plt.xlabel('Total amount in dollars')
plt.ylabel('Times a total amount is in the data')

plt.figure(figsize=(12,6))
ax = sns.histplot(data=df, x='total_amount', bins=range(-10,101,5))
ax.set_xticks(range(-10, 101, 5))
plt.title('Total amount histogram without outliers', fontsize=18)
plt.xlabel('Total amount in dollars')
plt.ylabel('Times a total amount is in the data')
```

**tip amount**

```python
# Create box plot of tip_amount
plt.figure(figsize=(15,3))
sns.boxplot(data=df, x='tip_amount', fliersize=1)
plt.title('Tip amount distribution', fontsize=18)
plt.xlabel('Tip amount in dollars')
```

```python
# Create histogram of tip_amount
plt.figure(figsize=(12,6))
sns.histplot(data=df, x='tip_amount', bins=range(0,201,5))
plt.title('Tip amount histogram', fontsize=18)
plt.xlabel('Tip amount in dollars')
plt.ylabel('Times a tip amount is in the data')

plt.figure(figsize=(12,6))
ax = sns.histplot(data=df, x='tip_amount', bins=range(0,21,1))
ax.set_xticks(range(0,21,1))
plt.title('Tip amount histogram without outliers', fontsize=18)
plt.xlabel('Tip amount in dollars')
plt.ylabel('Times a tip amount is in the data')
```

**tip_amount by vendor**

```python
# Create histogram of tip_amount by vendor
plt.figure(figsize=(12,6))
sns.histplot(data=df, x='tip_amount', bins=range(0,201,5), hue='VendorID', palette='bright', multiple='stack')
plt.title('Tip amount by vendor histogram', fontsize=18)

plt.figure(figsize=(12,6))
ax = sns.histplot(data=df, x='tip_amount', bins=range(0,21,1), hue='VendorID', palette='bright', multiple='stack')
ax.set_xticks(range(0,21,1))
plt.title('Tip amount by vendor histogram without outliers', fontsize=18)
```

**Mean tips by passenger count**

Examine the unique values in the `passenger_count` column.

```python
df['passenger_count'].value_counts()
```

```python
# Calculate mean tips by passenger_count
df_tip_pass = df.groupby(['passenger_count'])[['tip_amount']].mean()
df_tip_pass
```

```python
# Create bar plot for mean tips by passenger count
plt.figure(figsize=(16,8))
data=df_tip_pass.tail(-1).reset_index()
colores = sns.color_palette('Purples', len(data))
sns.barplot(data=data,
            x='passenger_count',
            y='tip_amount',  
            palette=colores
            )
plt.title('Mean tip mount by passenger count', fontsize=18)
plt.xlabel('Quantity of passengers')
plt.ylabel('Tip amount')
```

**Create month and day columns**

```python
# Create a month column
df['month'] = df['tpep_pickup_datetime'].dt.month_name()

# Create a day column
df['day'] = df['tpep_pickup_datetime'].dt.day_name()
```

**Plot total ride count by month**

Begin by calculating total ride count by month.

```python
# Get total number of rides for each month
monthly_rides = df['month'].value_counts()
monthly_rides
```

Reorder the results to put the months in calendar order.

```python
# Reorder the monthly ride list so months go in order
month_order = [
    'January', 'February', 'March', 'April', 'May', 'June',
    'July', 'August', 'September', 'October', 'November', 'December'
]

monthly_rides = monthly_rides.reindex(month_order)
```

```python
# Show the index
monthly_rides.index
```

```python
# Create a bar plot of total rides per month
plt.figure(figsize=(16,8))
ax = sns.barplot(x=monthly_rides.index,y=monthly_rides)
ax.set_xticklabels(month_order)
plt.title('Total rides per month',fontsize=18)
plt.xlabel('Month')
plt.ylabel('Total rides')
```

**Plot total ride count by day**

Repeat the above process, but now calculate the total rides by day of the week.

```python
# Repeat the above process, this time for rides by day
weekly_rides = df['day'].value_counts()
week_order = [
    'Monday', 'Tuesday', 'Wednesday', 'Thursday', 
    'Friday', 'Saturday', 'Sunday'
]

weekly_rides = weekly_rides.reindex(week_order)
weekly_rides.index
```

```python
# Create bar plot for ride count by day
plt.figure(figsize=(12,6))
ax = sns.barplot(x=weekly_rides.index,y=weekly_rides)
ax.set_xticklabels(week_order)
plt.title('Total rides per day', fontsize=18)
plt.xlabel('Day')
plt.ylabel('Total rides')
```

**Plot total revenue by day of the week**

Repeat the above process, but now calculate the total revenue by day of the week.

```python
# Repeat the process, this time for total revenue by day
revenue_day = df.groupby('day')[['total_amount']].sum()
revenue_day = revenue_day.reindex(week_order)
revenue_day
```

```python
# Create bar plot of total revenue by day
plt.figure(figsize=(12, 6))
ax = sns.barplot(
    data=revenue_day, 
    x=revenue_day.index, 
    y='total_amount'
)
plt.title('Total revenue per day', fontsize=18)
plt.xlabel('Day')
plt.ylabel('Total revenue')
```

**Plot total revenue by month**

```python
# Repeat the process, this time for total revenue by month
revenue_month = df.groupby('month')[['total_amount']].sum()

revenue_month = revenue_month.reindex(month_order)
revenue_month
```

```python
# Create a bar plot of total revenue by month
plt.figure(figsize=(12, 6))
ax = sns.barplot(
    data=revenue_month, 
    x=revenue_month.index, 
    y='total_amount'
)
plt.title('Total revenue per month', fontsize=18)
plt.xlabel('Month')
plt.ylabel('Total revenue')
```

## BONUS CONTENT

To confirm your conclusion, consider the following experiment:
1. Create a sample of coordinates from a normal distribution&mdash;in this case 1,500 pairs of points from a normal distribution with a mean of 10 and a standard deviation of 5
2. Calculate the distance between each pair of coordinates 
3. Group the coordinates by endpoint and calculate the mean distance between that endpoint and all other points it was paired with
4. Plot the mean distance for each unique endpoint

```python
#BONUS CONTENT

#1. Generate random points on a 2D plane from a normal distribution
points = np.round(np.random.normal(loc=10, scale=5, size=(3000, 2)), 1) #Generates 3,000 points (X,Y) with mean=10, std=5 and rounded them to 1 decimal
start, end = np.split(points, 2) #Split the matrix in two, 1500 points for PU(start) and 1500 for DO(end)

# 2. Calculate Euclidean distances between points in first half and second half of array
distances = np.sqrt(np.sum((start - end)**2, axis=1))

# 3. Group the coordinates by "drop-off location", compute mean distance
df_bonus = pd.DataFrame({ # Creating a dataframe with all start and end points, and the distance.
    'start': list(map(tuple, start)), #Convert each pair to a tuple, so the (X, Y) points are inmutable.
    'end': list(map(tuple, end)),
    'distance': distances
})

mean_distance = df_bonus.groupby('end')['distance'].mean().sort_values().to_frame() # Groups by DO and calculate the mean and sorted in ascending order, getting back a dataframe

# 4. Plot the mean distance between each endpoint ("drop-off location") and all points it connected to
plt.figure(figsize=(16, 6))
ax = sns.barplot(x=mean_distance.index, y=mean_distance['distance'])

plt.title('Mean distance by drop-off location (Normal Distribution Simulation)', fontsize=16)
plt.xlabel('Drop-off Location')
ax.set_xticks([]) #Removing all the values from x axe
plt.ylabel('Mean distance')
```
